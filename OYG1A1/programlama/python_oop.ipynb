## Örnek  — BankAccount (Banka Hesabı)

Hedef: para yatır/çek, bakiyeyi koru.
- Kural: negatif olmaz, fazla çekme olmaz.


````python
class BankAccount:
    def __init__(self, owner, balance=0):
        if balance < 0:
            raise ValueError("Bakiye negatif olamaz")
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Pozitif yatır")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Pozitif çek")
        if amount > self.balance:
            raise ValueError("Yetersiz bakiye")
        self.balance -= amount

acc = BankAccount("Ali", 100)
acc.deposit(50)
acc.withdraw(20)
print(acc.owner, acc.balance)
````

## Örnek  — TodoList (Görev Listesi)

Hedef: görev ekle, tamamla, göster.
- Kural: boş görev eklenmez, yanlış index olmaz.


````python
class TodoList:
    def __init__(self, title):
        self.title = title
        self.tasks = []  # {"text":..., "done":...}

    def add(self, text):
        text = text.strip()
        if text == "":
            raise ValueError("Boş görev olmaz")
        self.tasks.append({"text": text, "done": False})

    def done(self, index):
        if not (0 <= index < len(self.tasks)):
            raise IndexError("Böyle bir görev yok")
        self.tasks[index]["done"] = True

    def show(self):
        print("==", self.title, "==")
        for i, t in enumerate(self.tasks):
           if t["done"]:
               mark = "[+]"
           else:
               mark = "[x]"
        print(i, mark, t["text"])

todo = TodoList("Bugün")
todo.add("Python çalış")
todo.add("Su iç")
todo.done(0)
todo.show()
````

##  Örnek — Inventory (Çanta) (Composition)

Hedef: çantaya eşya koy, kapasite dolarsa hata ver.


````python
class Item:
    def __init__(self, name, weight=1):
        if weight <= 0:
            raise ValueError("Ağırlık pozitif olmalı")
        self.name = name
        self.weight = weight

class Inventory:
    def __init__(self, capacity=10):
        self.capacity = capacity
        self.items = []

    def total_weight(self):
        sum = 0
        for it in self.items:
            sum += it.weight
        return sum
    def add(self, item):
        if self.total_weight() + item.weight > self.capacity:
            raise ValueError("Çanta dolu!")
        self.items.append(item)

    def show(self):
        print("== Envanter ==")
        for it in self.items:
            print("-", it.name, f"({it.weight} kg)")
        print(f"Toplam ağırlık:{self.total_weight()} // {self.capacity}")

bag = Inventory(capacity=5)
bag.add(Item("İksir", 2))
bag.add(Item("Kılıç", 3))
bag.show()
````

## Örnek  — Kalıtım (Inheritance)

Hedef: ortak şeyleri `Animal`a yaz, farklı sesi alt sınıflar söylesin.


````python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):
        return "Hav hav!"

class Cat(Animal):
    def speak(self):
        return "Miyav!"

pets = [Dog("Karabaş"), Cat("Mırmır")]
for p in pets:
    print(p.name, "->", p.speak())
````
