# 🧰 Python Koleksiyonları ve Metotları

Python'da veri yapıları (List, Tuple, Set, Dictionary) sadece veri saklamakla kalmaz; veriler üzerinde işlem yapmamızı sağlayan **sihirli araçlara (metotlara)** da sahiptir.

Bir veri yapısının yanına `.` (nokta) koyduğumuzda bu araçlara erişebiliriz. Hadi en önemlilerini inceleyelim!

## 1. 📋 Listeler (Lists) ve Metotları

Listeler en çok metoda sahip olan yapıdır. Ekleme, çıkarma, sıralama gibi birçok işlem yapabiliriz.

```python
liste = [10, 20, 30, 40]
print(f"Başlangıç: {liste}")

# --- EKLEME METOTLARI ---
liste.append(50)       # Sona ekler
liste.insert(1, 15)    # Belirli bir indekse (1. sıraya) ekler
liste.extend([60, 70]) # Başka bir listeyi ucuna ekler
print(f"Ekleme sonrası: {liste}")

# --- SİLME METOTLARI ---
liste.pop()            # En sondakini siler (veya indeks verilebilir)
liste.remove(30)       # Değeri bulup siler (İndeks değil, değer!)
print(f"Silme sonrası: {liste}")

# --- DÜZENLEME VE ARAMA ---
sayilar = [5, 1, 9, 2, 5]
sayilar.sort()         # Küçükten büyüğe sıralar
print(f"Sıralı: {sayilar}")

sayilar.reverse()      # Listeyi ters çevirir
print(f"Ters: {sayilar}")

kac_tane_5 = sayilar.count(5) # İçinde kaç tane 5 var?
print(f"5 sayısı kaç tane?: {kac_tane_5}")

# Hepsini temizleme
sayilar.clear()
print(f"Temizlendikten sonra: {sayilar}")
```

```python
# 5 elemanlı bir liste tanımlayınız.
l1=["ali","veli",35,True,58.8]
# 3. indisteki değeri 22 ile değiştirin
l1[3]=22
print(l1)
# bu listenin sonuna yeni 2 tam sayı değeri ekleyin
l1.extend([45,58])
print(l1)
# for döngüsü ile listenizin elemanlarını ekrana yazdırın
for i in l1:
    print(i)

# listedeki 4. elemanı siliniz.
e4=l1[4]
l1.remove(e4) 
print(l1)
# listenin sonuncu elemanın silip değerini bir değikene aktar(pop)
son=l1.pop()
print(l1,"son:",son)
# l2 isimli boş liste oluşturunuz.
l2=[]
l3=list()
print(type(l2),type(l3))
```

## 2. 🔒 Demetler (Tuples) ve Metotları

Tuple'lar **değiştirilemez** olduğu için ekleme/silme metotları (append, remove vs.) **YOKTUR**.
Sadece içindekileri sayabilir veya yerini bulabiliriz.

```python
notlar = (90, 80, 70, 90, 60, 90)


# .count() -> Eleman sayısını bulma
print(f"90 notu kaç kez alınmış?: {notlar.count(90)}")

# .index() -> Elemanın kaçıncı sırada olduğunu bulma
# Not: Bulduğu İLK indeksi verir.
print(f"70 notu kaçıncı sırada (indeks)?: {notlar.index(70)}")

#ikinci indisi ekrana yazdırın.
# notlar[2]=85
print(notlar[2])
```

## 3. 🎱 Kümeler (Sets) ve Metotları

Kümeler matematikteki kümeler gibidir. İndeks olmadığı için `append` yerine `add` kullanılır.

```python
kume = {1, 2, 3}
kume.add(3)
print("satır1",kume)
# --- EKLEME VE SİLME ---
kume.add(4)       # Tek bir eleman ekler
kume.update([5, 6, 7,1,7,2,3]) # Çoklu eleman ekler (Liste gönderilebilir)
print("update",kume)
kume.discard(2)   # Elemanı siler (Varsa siler, yoksa hata vermez - Güvenli)
kume.remove(3)    # Elemanı siler (Yoksa hata verir!)
print(f"Güncel Küme: {kume}")

# --- KÜME İŞLEMLERİ ---
a = {1, 2, 3,4,5}
b = {3, 4, 5,6,7}

print(f"Birleşim (union): {a.union(b)}")
print(f"Kesişim (intersection): {a.intersection(b)}")
print(f"Fark (difference - a'da olup b'de olmayan): {a.difference(b)}")
print("len:",len(kume))
print("len:",len("bilsem"))
```

## 4. 📖 Sözlükler (Dictionaries) ve Metotları

Sözlüklerde anahtarlar (keys) ve değerler (values) ile çalışmak için özel metotlar vardır.

```python
oyuncu = {
    "ad": "Süper Mario",
    "can": 3,
    "puan": 1500
}

# --- ERİŞİM METOTLARI ---
print(f"Sadece Anahtarlar: {oyuncu.keys()}")
print(f"Sadece Değerler: {oyuncu.values()}")
print(f"İkili Gruplar Halinde: {oyuncu.items()}")

# .get() -> Güvenli Erişim
# oyuncu["seviye"] yazarsak hata alırız çünkü yok.
# Ama .get() kullanırsak None döner veya varsayılan değer atayabiliriz.
seviye = oyuncu.get("seviye", "Başlangıç") # Seviye yoksa 'Başlangıç' yaz
oyuncu["cinsiyet"]="erkek"
oyuncu["can"]=2500
print(f"Oyuncu Seviyesi: {seviye}")

# --- GÜNCELLEME VE SİLME ---
oyuncu.update({"can": 5, "hiz": "Normal"}) # Varsa günceller, yoksa ekler
silinen_puan = oyuncu.pop("puan") # Puanı siler ve değerini bize verir

print(f"Son Durum: {oyuncu}")
print(f"Silinen Puan: {silinen_puan}")
```
